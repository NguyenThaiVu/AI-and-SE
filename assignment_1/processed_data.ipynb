# Description

In this notebook, I will explore the dataset

```python
import os 
import numpy as np 
import pandas as pd 
```

```python
PATH_RAW_DATA = os.path.join(os.getcwd(), 'dataset', 'raw')
PATH_PROCESSED_DATA = os.path.join(os.getcwd(), 'dataset', 'processed')
PATH_FILE_PROCESSED = os.path.join(PATH_PROCESSED_DATA, 'processed_data.csv')
```

# 1. Read raw dataset

```python
list_file_raw = os.listdir(PATH_RAW_DATA)
list_file_raw = [filename for filename in list_file_raw if filename.endswith('.csv')]

print(f"Number of raw data files: {len(list_file_raw)}")
```

```python
df = pd.DataFrame()
for filename in list_file_raw:
    try:
        file_path = os.path.join(PATH_RAW_DATA, filename)
        temp_df = pd.read_csv(file_path)
        df = pd.concat([df, temp_df], ignore_index=True)
    except Exception as e:
        print(f"Error reading {filename}: {e}")
        continue
    
print(f"Shape of combined raw data: {df.shape}")
df.head()
```

```python
df.to_csv(PATH_FILE_PROCESSED, index=False)
print(f"Processed data saved to {PATH_FILE_PROCESSED}")
```
